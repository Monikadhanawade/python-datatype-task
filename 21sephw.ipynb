 complex :Stores complex numbers with real and imaginary parts.

```python
c=False
print(c)
print(type(c))
```

Boolean: Stores logical values: True or False.

```python
a=89.00
print(a)
print(type(a))
```

float: Stores numbers with decimal points.

```python
age=10
print(age)
print(type(age))
```

int: Stores whole numbers without decimal points.

```python
name="Monika"
print(name)
print(type(name))
```

string: Stores text or a sequence of characters.

1. Single quotes ' '
Use them for a normal, single-line string.

name = 'John'
print(name)

Output:John

2. Double quotes " "
They work exactly like single quotes.

name = "John"
print(name)

You can choose either ' ' or " " based on convenience.
